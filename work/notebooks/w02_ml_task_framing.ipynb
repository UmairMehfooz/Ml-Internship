## Success Metric

Since this is a classification task, accuracy alone is not enough.

More useful metrics include:

- Precision
- Recall
- F1 Score

Recall is especially important because missing a declining page means traffic may continue to fall.

A good model should identify as many truly declining pages as possible while avoiding unnecessary refreshes.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

## Why ML Beats a Fixed Rule

A simple rule such as

"If CTR is below 2%, refresh the page."

would ignore many other factors.

Search performance depends on multiple variables working together, including:

- search volume
- average ranking
- content age
- CTR
- impressions

A page with low CTR may still perform well if it ranks highly or has very high search demand.

Machine learning can discover these complex relationships and produce better recommendations than a single threshold.
